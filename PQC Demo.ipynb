We will demonstrate a key exchange using Kyber key encapsulation, which is based on lattice cryptography.

0. Import module-lattice key encapsulation mechanism (ML-KEM) using the kyber-py library.

```python
from kyber_py.ml_kem import ML_KEM_512
```

1. Alice generates a public key and private key.

```python
public_key, private_key = ML_KEM_512.keygen()
```

2. Alice sends public key to Bob.

3. Bob **encapsulates** public key to get shared key and cipher text.

```python
bob_shared_key, ciphertext = ML_KEM_512.encaps(public_key)
```

4. Bob sends ciphertext to Alice.

5. Alice **decapsulates** ciphertext using private key.

```python
alice_shared_key = ML_KEM_512.decaps(private_key, ciphertext)
```

Moment of truth... Does Alice's key match Bob's key?

```python
if alice_shared_key == bob_shared_key:
    print("Success! Alice and Bob successfully shared a key in a quantum-resistant manner.")
else:
    print("Something went wrong...")
```
